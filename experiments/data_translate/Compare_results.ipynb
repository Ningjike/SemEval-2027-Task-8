# Qwen3 与 Aya 结果对比

```python
import json
from pathlib import Path
import pandas as pd

RUN_ROOT = Path("/content/drive/MyDrive/sft_v5/model_runs")
MODELS = ["qwen3_8b", "aya_expanse_8b"]
configs = [json.loads((RUN_ROOT / name / "run_config.json").read_text()) for name in MODELS]
for key in ["data_sha256", "epochs", "max_length", "seed", "effective_batch_size", "learning_rate", "optimizer_steps"]:
    assert configs[0][key] == configs[1][key], f"两次运行的 {key} 不一致"

rows = []
for name in MODELS:
    metrics = json.loads((RUN_ROOT / name / "test_metrics.json").read_text())
    for language, values in metrics["by_language"].items():
        rows.append({"model": name, "language": language, **values})
pd.DataFrame(rows).set_index(["model", "language"])
```
